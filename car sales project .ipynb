**outliers**


```python
# Annual income
q1=df["Annual Income"].quantile(0.25)
q3=df["Annual Income"].quantile(0.75)
IQR=q3-q1
print("IQR:",IQR)
l_bound=q1-(1.5*IQR)
u_bound=q3+(1.5*IQR)
outliers=df[(df["Annual Income"]<l_bound)|(df["Annual Income"]>u_bound)]
print("outliers below lower_limit:",len(df[df["Annual Income"]<l_bound]))
print("outliers above upper_limit:",len(df[df["Annual Income"]>u_bound]))
outliers
```

- Here we observe annual income column have 816 records of outliers 

```python
q1=df["Price ($)"].quantile(0.25)
q3=df["Price ($)"].quantile(0.75)
IQR=q3-q1
print("IQR:",IQR)
l_bound=q1-(1.5*IQR)
u_bound=q3+(1.5*IQR)
outliers=df[(df["Price ($)"]<l_bound)|(df["Price ($)"]>u_bound)]
print("outliers below lower_limit:",len(df[df["Price ($)"]<l_bound]))
print("outliers above upper_limit:",len(df[df["Price ($)"]>u_bound]))
outliers
```

```python
# Price_Category: categorizes car prices into quartiles (Low, Medium, High, Premium)
df1["Price_category"]=pd.qcut(df1["Price ($)"],q=4,labels=("Low","Medium",'High',"Premium"))
```

```python
df1.head()
```

```python
price_bins = pd.qcut(df['Price ($)'], q=4, retbins=True)
bins = price_bins[1] 
print(bins)
```

```python
# Income_Group: categorizes customers' annual income into quartiles (Low, Medium, High, Very_High)
df1["Income_Group"]=pd.qcut(df["Annual Income"], q=4, labels=("Low","Medium","High","Very_High"))
```

```python
df1.head()
```

```python
income_bins=pd.qcut(df["Annual Income"],q=4, retbins=True)
bins=income_bins[1]
bins = np.round(bins).astype(int)
bins
```

```python
# Customer_ID_Flag: indicates whether a customer purchased more than once (1 = repeated, 0 = single purchase)
df1["Customer_ID_Flag"]= df1.duplicated(subset=['Customer Name'], keep=False)
df1['Customer_ID_Flag'] = df1['Customer_ID_Flag'].astype(int)
```

```python
df1.tail()
```

```python
# Extract numeric quarter (1–4)
df1['Quarter_Number'] = df1['Date'].dt.quarter

# Combine Year + Quarter (like "2023Q1")
df1['Quarter_Year'] = df1['Date'].dt.to_period('Q').astype(str)
df1.head()
```

```python
df1=df1.drop(columns=["Car_id","Customer Name","Dealer_No "],axis=1)
```

**Note**
- Income_Group: This column categorizes customers' annual income into four quartiles — Low, Medium, High, and Very High — based on the distribution of their income values.
-  Price_Category: Categorizes car prices into four quartiles — Low, Medium, High, and Premium — based on the distribution of car price values.
-  Customer_ID_Flag: Indicates whether a customer has made more than one purchase — 1 for repeated customers and 0 for single-purchase customers.
-  Date: date column divided into quarters.
-  also remove some unimportant columns

*Fluctuating Price Trend:*
- Across all regions, the car prices fluctuate from 2022Q1 to 2023Q4, indicating some seasonal or quarterly variation.

*No Strong Upward or Downward Trend:*
- Overall, there is no consistent increase or decrease in average car prices over time — prices remain relatively stable.

*Significant Dip in 2023Q1:*
- Most regions, especially Pasco, Greenville, and Janesville, show a sharp decline in average price during 2023Q1, followed by a recovery afterward.

*Peak in Mid-2022:*
- Regions such as Aurora and Middletown reached their highest price points in 2022Q3, suggesting a strong sales or high-value period during that quarter.

*Regional Variation:*

- Pasco shows the most volatility, with steep rises and falls.

- Austin and Scottsdale remain relatively stable, showing smoother price movements.

- Aurora often stays among the higher average prices compared to other regions.
